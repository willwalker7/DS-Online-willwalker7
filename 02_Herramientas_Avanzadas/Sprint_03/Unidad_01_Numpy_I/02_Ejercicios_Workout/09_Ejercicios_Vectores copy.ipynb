### Ejercicio 3



Utiliza el algoritmo visto en las pildoras del workout para determinar las clases extaordinarias que corresponden a Teresa y Udo.

````python
import math

def distancia_2d(vec1, vec2, precision = 2):
    suma = 0
    for i in range(2):
        suma += (vec1[i] - vec2[i])**2   
    distancia = math.sqrt(suma)
    distancia = round(distancia, precision)
    return distancia

vec_teresa = alumnos_aidone["Teresa"]
vec_udo = alumnos_aidone["Udo"]

for alumno_ref in ["Rodrigo", "Gabriela", "Mateo"]:
    vec_ref = alumnos_aidone[alumno_ref]
    distancia = distancia_2d(vec_teresa, vec_ref, 1)
    print(f"La distancia de Teresa con {alumno_ref} es de {distancia}")
    
for alumno_ref in ["Rodrigo", "Gabriela", "Mateo"]:
    vec_ref = alumnos_aidone[alumno_ref]
    distancia = distancia_2d(vec_udo, vec_ref, 1)
    print(f"La distancia de Udo con {alumno_ref} es de {distancia}")
````
