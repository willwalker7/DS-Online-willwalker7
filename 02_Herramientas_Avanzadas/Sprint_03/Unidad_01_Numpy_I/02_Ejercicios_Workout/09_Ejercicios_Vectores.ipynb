### Ejercicio 4



Modifica la función de cálculo de distancias que hicimos en el workout para que funcione con vectores de cualquier dimensión

````python
def distancia(vec1, vec2, precision=2):
    if len(vec1) != len(vec2):
        raise ValueError("Los vectores deben tener la misma dimensión")
    
    import math
    suma = sum((a - b)**2 for a, b in zip(vec1, vec2))
    return round(math.sqrt(suma), precision)
````

### Ejercicio 7



Haz un pequeño programa que caclcule el coseno del ángulo entre cada pareja de vectores utilizando las funciones que desarrollamos en el workout. No hagas las operaciones una a una.

````python
def dot_product(vec1, vec2, precision = 2):
    producto = 0
    for indice in range(len(vec1)): # Importante sólo podemos hacer producto escalar de vectores con el mismo número de componentes o dimensiones
        producto += vec1[indice] * vec2[indice]
    producto = round(producto, precision)
    return producto
````

````python
def norm_vec(vec, precision = 1):
    escalar = dot_product(vec,vec) # Esta es la magia de las funciones y los métodos construir unos sobre otros :-), reusar, craft, craft
    resultado = math.sqrt(escalar)
    resultado = round(resultado, precision)
    return resultado
````
