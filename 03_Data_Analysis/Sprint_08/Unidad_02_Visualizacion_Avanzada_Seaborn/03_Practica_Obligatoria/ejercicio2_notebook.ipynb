```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
df = pd.read_csv("./data/titanic.csv")
```

```python
data = df["alive"].value_counts()

fig,ax = plt.subplots(1,1,figsize = (4,4))

ax.pie(data,
       labels = ["No", "Sí"],
       autopct = '%.2f%%',
       startangle = 90);
ax.set_title("Supervivientes");
```
